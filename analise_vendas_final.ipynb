# 📊 Análise de Vendas

Este projeto tem como objetivo analisar dados de vendas utilizando Python,
com foco em aprendizado prático e construção de portfólio profissional
para vagas de Analista de Dados Júnior (Home‑Office).

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('data/vendas.csv')
df.head()
```

```python
df.info()
```

## 📋 Entendimento dos Dados

O conjunto de dados contém informações de vendas, incluindo data,
produto, categoria e valor. Os dados são simples e adequados
para análises iniciais.

## ❓ Perguntas de Negócio

- Como as vendas evoluíram ao longo do tempo?
- Quais produtos geram maior faturamento?

```python
df['data_venda'] = pd.to_datetime(df['data_venda'])
```

```python
vendas_por_data = df.groupby('data_venda')['valor'].sum()
vendas_por_data
```

```python
plt.figure(figsize=(10, 5))
plt.plot(vendas_por_data.index, vendas_por_data.values, marker='o')

plt.title('Evolução das Vendas ao Longo do Tempo')
plt.xlabel('Data')
plt.ylabel('Valor Total de Vendas (R$)')
plt.grid(True)

plt.show()
```

📌 O gráfico mostra a variação das vendas ao longo do tempo,
permitindo identificar períodos de maior ou menor faturamento.

```python
vendas_por_produto = (
    df.groupby('produto')['valor']
    .sum()
    .sort_values(ascending=False)
)

vendas_por_produto
```

```python
top_produtos = vendas_por_produto.head(5)

plt.figure(figsize=(8, 4))
plt.bar(top_produtos.index, top_produtos.values)

plt.title('Top 5 Produtos por Valor de Vendas')
plt.xlabel('Produto')
plt.ylabel('Valor Total de Vendas (R$)')
plt.xticks(rotation=45)

plt.show()
```

📌 O gráfico evidencia que poucos produtos concentram a maior parte do faturamento,
indicando oportunidades de foco estratégico e otimização do portfólio de vendas.

## ✅ Conclusões Finais

- As vendas apresentam variação ao longo do tempo, permitindo identificar períodos de maior faturamento
- Um pequeno grupo de produtos concentra a maior parte da receita
- A análise pode apoiar decisões estratégicas de foco comercial e planejamento de vendas
